# NN Layers - 2D Convolution, Pooling and Padding

    1.1.  torch.nn.Conv2d          - Applies a 2D convolution
    1.2.  torch.nn.ConvTranspose2d - Applies a 2D transposed convolution
    1.3.  torch.nn.MaxPool2d           - Applies a 2D max pooling.
    1.4.  torch.nn.MaxUnpool2d         - Computes a partial inverse of MaxPool2d
    1.5.  torch.nn.AvgPool2d           - Applies a 2D average pooling
    1.6.  torch.nn.FractionalMaxPool2d - Applies a 2D fractional max pooling
    1.7. torch.nn.AdaptiveMaxPool2d   - Applies a 2D adaptive max pooling
    1.8. torch.nn.AdaptiveAvgPool2d   - Applies a 2D adaptive average pooling
    1.9.  torch.nn.ReflectionPad2d - Pads the input tensor using the reflection of the input boundary.
    1.10.  torch.nn.ReplicationPad2d - Pads the input tensor using the replication of the input boundary.
    1.11.  torch.nn.ZeroPad2d - Pads the input tensor boundaries with zero.
    1.12.  torch.nn.ConstantPad2d - Pads the input tensor boundaries with a constant value.
    

Now, we take a look at the same features discussed in the last module, but in two dimensions.

```python
import os
import sys
import torch
import numpy as np
import torch.nn as nn
from torchvision import transforms, datasets
from PIL import Image
import cv2
import matplotlib.pyplot as plt
import torchvision
```

```python
# Input
!wget -O dog.jpg https://cdn.britannica.com/79/232779-050-6B0411D7/German-Shepherd-dog-Alsatian.jpg
image_name = "dog.jpg"
image_pil = Image.open(image_name)

transform = transforms.Compose([transforms.ToTensor()])
image = transform(image_pil).float()
image_nchw = torch.unsqueeze(image, 0)
print(image_nchw.size())
plt.imshow(image_pil)
```

#Pooling

## Visualizing Conv2d


```python
conv_layer = nn.Conv2d(in_channels=3, out_channels=3, kernel_size=3, stride=2)
output_tensor = conv_layer(image_nchw)

# Convert the output tensor to a PIL image
output_squeezed = torch.squeeze(output_tensor)
transform_to_pil = transforms.Compose([transforms.ToPILImage(mode=None)])
convoluted_pil = transform_to_pil(output_squeezed)

# Create a figure with two subplots
fig, ax = plt.subplots(1, 2, figsize=(14, 7)) # Increased figure size for better label visibility

# Display the original image
ax[0].imshow(image_pil)
ax[0].set_title('Original Image')
ax[0].set_xlabel(f'Width: {image_pil.size[0]} pixels')
ax[0].set_ylabel(f'Height: {image_pil.size[1]} pixels')
ax[0].axis('on') # Keep axes on to show labels

# Display the convoluted image
ax[1].imshow(convoluted_pil)
ax[1].set_title('Conv2d Output')
ax[1].set_xlabel(f'Width: {convoluted_pil.size[0]} pixels')
ax[1].set_ylabel(f'Height: {convoluted_pil.size[1]} pixels')
ax[1].axis('on') # Keep axes on to show labels

# Adjust subplot parameters for a tight layout and display the plots
plt.tight_layout()
plt.show()
```
