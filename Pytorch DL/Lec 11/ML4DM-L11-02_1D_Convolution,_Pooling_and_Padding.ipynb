# 1D Convolution





We generate a sample 1D signal using `torch.randn` and plot it to visualize the initial state. This signal will serve as the input for all subsequent operations.


```python
import torch
import matplotlib.pyplot as plt

# 1. Generate a 1D input signal named input_signal_1d
# Using batch_size=1, channels=1, length=50 as suggested for example
input_signal_1d = torch.randn(1, 1, 500)

# 2. Extract the actual signal for plotting
# Select the first batch item and first channel, then convert to NumPy array
plot_signal = input_signal_1d[0, 0, :].detach().numpy()

# 3. Plot the NumPy array
plt.figure(figsize=(10, 4))
plt.plot(plot_signal)
plt.title('Generated 1D Input Signal')
plt.xlabel('Length')
plt.ylabel('Value')
plt.grid(False)
plt.show()

print("Generated 1D input signal with shape:", input_signal_1d.shape)
```

Now that the 1D input signal is generated, we will define a `torch.nn.Conv1d` layer and apply it to the `input_signal_1d`. We will then extract and plot the output signal from the convolution.



```python
import torch.nn as nn
import matplotlib.pyplot as plt

# 4. Define a 1D Convolutional layer
# Using in_channels=1 (from input_signal_1d), out_channels=1, kernel_size=3, stride=1
conv1d_layer = nn.Conv1d(in_channels=1, out_channels=1, kernel_size=3, stride=1, padding=0)

# 5. Apply the convolution to the generated input signal
output_signal_1d = conv1d_layer(input_signal_1d)

print("Input signal shape:", input_signal_1d.shape)
print("Output signal shape after Conv1d:", output_signal_1d.shape)

# 6. Extract the signals for plotting
plot_input_signal = input_signal_1d[0, 0, :].detach().numpy()
plot_output_signal = output_signal_1d[0, 0, :].detach().numpy()

# 7. Plot both the original and convolved signals
plt.figure(figsize=(12, 6))
plt.plot(plot_input_signal, label='Original 1D Signal')
plt.plot(plot_output_signal, label='Convolved 1D Signal')
plt.title('Original vs. Convolved 1D Signal')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)
plt.show()
```

## Effect of Stride on 1D Convolution

This section visualizes the impact of different `stride` values on the output of a `torch.nn.Conv1d` operation. A larger stride leads to more aggressive downsampling, reducing the length of the output signal.

```python
import torch.nn as nn
import matplotlib.pyplot as plt

# Define 1D Convolutional layers with different strides
conv1d_stride1 = nn.Conv1d(in_channels=1, out_channels=1, kernel_size=3, stride=1, padding=0)
conv1d_stride2 = nn.Conv1d(in_channels=1, out_channels=1, kernel_size=3, stride=2, padding=0)

# Apply convolutions to the generated input signal
output_stride1 = conv1d_stride1(input_signal_1d)
output_stride2 = conv1d_stride2(input_signal_1d)

print("Input signal shape:", input_signal_1d.shape)
print("Output shape with stride=1:", output_stride1.shape)
print("Output shape with stride=2:", output_stride2.shape)

# Extract signals for plotting
plot_input_signal = input_signal_1d[0, 0, :].detach().numpy()
plot_output_stride1 = output_stride1[0, 0, :].detach().numpy()
plot_output_stride2 = output_stride2[0, 0, :].detach().numpy()

# Plotting
plt.figure(figsize=(16, 8))

plt.subplot(2, 1, 1) # Two rows, one column, first plot
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.plot(plot_output_stride1, label='Conv1d (Stride=1) Output', alpha=0.7)
plt.title('Original vs. Convolved 1D Signal (Stride=1)')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)

plt.subplot(2, 1, 2) # Two rows, one column, second plot
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.plot(plot_output_stride2, label='Conv1d (Stride=2) Output', alpha=0.7)
plt.title('Original vs. Convolved 1D Signal (Stride=2)')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)

plt.tight_layout()
plt.show()
```

### Observation on Stride Effect

From the plots, we can observe the following effects of stride:

*   **Stride = 1**: When `stride=1`, the convolution kernel moves one step at a time. This results in an output signal that is only slightly shorter than the input signal (due to `kernel_size > 1` and `padding=0`). The overall shape and details of the signal are largely preserved, though transformed by the convolution weights.

*   **Stride = 2**: When `stride=2`, the convolution kernel skips one position for each step. This significantly downsamples the signal, resulting in an output signal that is approximately half the length of the input signal. This downsampling effect makes the output a more compressed representation, capturing broader features and discarding finer details from the original signal.

In summary, increasing the stride value reduces the output size and provides a coarser-grained representation of the input signal, which is useful for reducing computational load and extracting high-level features.

## Apply and Plot ConvTranspose1d



Let us define a `torch.nn.ConvTranspose1d` layer, apply it to the `input_signal_1d`, and then plot both the original and the transposed convolved signals for visual comparison to highlight the upsampling or expansion effect.
.



```python
import torch.nn as nn
import matplotlib.pyplot as plt

# 1. Define a 1D Transposed Convolutional layer
# Using in_channels=1, out_channels=1, kernel_size=3, stride=2, padding=0, output_padding=0
transposed_conv1d_layer = nn.ConvTranspose1d(in_channels=1, out_channels=1, kernel_size=3, stride=2, padding=0, output_padding=0)

# 2. Apply the transposed convolution to the generated input signal
output_transposed_signal_1d = transposed_conv1d_layer(input_signal_1d)

# 3. Print the shape of the input_signal_1d and the output of the ConvTranspose1d layer
print("Input signal shape:", input_signal_1d.shape)
print("Output signal shape after ConvTranspose1d:", output_transposed_signal_1d.shape)

# 4. Extract the signals for plotting
plot_input_signal = input_signal_1d[0, 0, :].detach().numpy()
plot_output_transposed_signal = output_transposed_signal_1d[0, 0, :].detach().numpy()

# 5 & 6. Plot both the original and transposed convolved signals
plt.figure(figsize=(14, 6))
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.plot(plot_output_transposed_signal, label='Transposed Convolved 1D Signal', alpha=0.7)
plt.title('Original vs. Transposed Convolved 1D Signal (Upsampling Effect)')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)
plt.show()
```

### Effect of 1D Transposed Convolution on the Signal

Based on the visualizations of the input 1D signal and the output of the 1D Transposed Convolution, we can observe the following effects:

1.  **Signal Length Expansion**: The most prominent effect is the expansion of the signal's length. The input signal had a length of 500, while the output signal from the `F.conv_transpose1d` operation has a length of 1000. This is characteristic of transposed convolutions, which are often used for upsampling or increasing the dimensionality of the input.

2.  **Increased Complexity/Detail**: The output signal appears to have more 'detail' or higher frequency components compared to the input. While the input signal shows some discernible patterns, the transposed convolution introduces a richer and potentially more complex structure.

3.  **Amplitude Changes**: The range of values (amplitude) in the output signal might differ significantly from the input. Transposed convolutions effectively 'spread out' the input information across a larger output space, and the specific weights used in the convolution determine how these amplitudes are scaled and distributed.

4.  **Pattern Transformation**: The original peaks and troughs in the input signal are transformed in the output. They are not simply scaled or shifted, but rather convoluted with the kernel, leading to new, broader, or more complex patterns. This demonstrates the operation's ability to 'deconvolve' or reconstruct a higher-resolution signal from a lower-resolution one, albeit often in a learned, non-exact way.

In summary, the 1D Transposed Convolution acts as an upsampling mechanism, increasing the length of the 1D signal and transforming its patterns and amplitude distribution through a learnable convolution process.

## Apply and Plot MaxPool1d


Imagine you have a series of numbers, or a grid of values from an image. Max pooling slides a small window (or filter) across this data and, at each step, picks out the largest value found inside the window. It does this over the whole input. Why do we need it?


* Reduces size: Max pooling makes your data smaller, which helps your network run faster and use less memory.
* Keeps important features: By only keeping the maximum values, max pooling helps focus on the most noticeable or important parts of the data, like strong edges in images or spikes in signals.
* Adds robustness: Since it only cares about the biggest value in each window, max pooling helps the network ignore small noise or irrelevant details.




To perform the max pooling operation, we will first define a `torch.nn.MaxPool1d` layer and apply it to the previously generated `input_signal_1d`. Then we will plot both the input and the resulting pooled output to visualize the downsampling effect and feature extraction characteristic of max pooling.



```python
import torch.nn as nn
import matplotlib.pyplot as plt

# 1. Define a 1D Max Pooling layer
# Using kernel_size=2, stride=2 for downsampling by a factor of 2
maxpool1d_layer = nn.MaxPool1d(kernel_size=2, stride=2)

# 2. Apply the max pooling layer to the generated input signal
output_pooled_signal_1d = maxpool1d_layer(input_signal_1d)

# 3. Print the shapes of both input and output signals
print("Input signal shape:", input_signal_1d.shape)
print("Output signal shape after MaxPool1d:", output_pooled_signal_1d.shape)

# 4. Extract the signals for plotting
plot_input_signal = input_signal_1d[0, 0, :].detach().numpy()
plot_output_pooled_signal = output_pooled_signal_1d[0, 0, :].detach().numpy()

# 5 & 6. Plot both the original and max-pooled signals
plt.figure(figsize=(14, 6))
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.plot(plot_output_pooled_signal, label='Max-Pooled 1D Signal', alpha=0.7)
plt.title('Original vs. Max-Pooled 1D Signal (Downsampling Effect)')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)
plt.show()
```

## Effect of Stride on `nn.MaxPool1d`

This section demonstrates how different `stride` values in `torch.nn.MaxPool1d` affect the length and characteristics of the output signal. A larger stride leads to more significant downsampling.

```python
import torch.nn as nn
import matplotlib.pyplot as plt

# Define MaxPool1d layers with different strides
# Using kernel_size=2 for all, to focus on stride effect
maxpool1d_stride2 = nn.MaxPool1d(kernel_size=2, stride=2)
maxpool1d_stride4 = nn.MaxPool1d(kernel_size=2, stride=4)

# Apply max pooling with different strides to the generated input signal
output_pooled_stride2 = maxpool1d_stride2(input_signal_1d)
output_pooled_stride4 = maxpool1d_stride4(input_signal_1d)

print("Input signal shape:", input_signal_1d.shape)
print("Output shape with stride=2:", output_pooled_stride2.shape)
print("Output shape with stride=4:", output_pooled_stride4.shape)

# Extract signals for plotting
plot_input_signal = input_signal_1d[0, 0, :].detach().numpy()
plot_output_pooled_stride2 = output_pooled_stride2[0, 0, :].detach().numpy()
plot_output_pooled_stride4 = output_pooled_stride4[0, 0, :].detach().numpy()

# Plotting
plt.figure(figsize=(16, 12))

plt.subplot(3, 1, 1) # Three rows, one column, first plot
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.title('Original 1D Signal')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)

plt.subplot(3, 1, 2) # Three rows, one column, second plot
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.plot(plot_output_pooled_stride2, label='MaxPool1d (Stride=2) Output', alpha=0.7)
plt.title('Original vs. MaxPool1d (Stride=2) Output')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)

plt.subplot(3, 1, 3) # Three rows, one column, third plot
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.plot(plot_output_pooled_stride4, label='MaxPool1d (Stride=4) Output', alpha=0.7)
plt.title('Original vs. MaxPool1d (Stride=4) Output')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)

plt.tight_layout()
plt.show()
```

### Observation on Stride Effect in `nn.MaxPool1d`

From the plots, we can clearly see the impact of increasing the `stride` value in `nn.MaxPool1d`:

*   **Stride = 2**: When `stride=2` (with `kernel_size=2`), the Max Pooling operation downsamples the signal by a factor of approximately 2. This means that for every two elements in the input, one maximum value is selected and passed to the output. The output signal's length is reduced significantly, but the overall shape of the dominant features is still somewhat preserved, albeit at a lower resolution.

*   **Stride = 4**: When `stride=4` (with `kernel_size=2`), the downsampling is even more aggressive. The Max Pooling operation skips more elements between each pooling window, leading to an output signal that is much shorter, approximately a quarter of the original length. This results in a much coarser representation of the input signal, capturing only the most prominent peaks over larger intervals.

In summary, a larger `stride` in `nn.MaxPool1d` results in a greater reduction in the signal's length and a more abstract, downsampled representation, which is effective for extracting high-level features and reducing computational load further down the network.

## Apply and Plot MaxUnpool1d

MaxUnpool1D takes the smaller, pooled data and places the maximum values back in their original positions, filling the other spots with zeros. It uses information saved during the pooling step to know exactly where to put these values.


To demonstrate the MaxPool1d and MaxUnpool1d operations, we will define both layers, apply them sequentially to the `input_signal_1d` (first MaxPool1d with return_indices=True, then MaxUnpool1d) and print the resulting tensor shapes. Finally, we plot the original, pooled, and unpooled signals for visual comparison.



```python
import torch.nn as nn
import matplotlib.pyplot as plt

# 1. Define a 1D Max Pooling layer and apply it
# Using kernel_size=2, stride=2 for downsampling by a factor of 2, and return_indices=True
maxpool1d_layer_unpool = nn.MaxPool1d(kernel_size=2, stride=2, return_indices=True)
pooled_data, indices = maxpool1d_layer_unpool(input_signal_1d)

# 2. Define a 1D Max Unpooling layer and apply it
# Using kernel_size=2, stride=2 corresponding to the pooling operation
maxunpool1d_layer = nn.MaxUnpool1d(kernel_size=2, stride=2)
unpooled_data = maxunpool1d_layer(pooled_data, indices, output_size=input_signal_1d.size())

# 3. Print the shapes of input, pooled, and unpooled signals
print("Input signal shape:", input_signal_1d.shape)
print("Pooled data shape:", pooled_data.shape)
print("Unpooled data shape:", unpooled_data.shape)

# 4. Extract the signals for plotting
plot_input_signal = input_signal_1d[0, 0, :].detach().numpy()
plot_pooled_signal = pooled_data[0, 0, :].detach().numpy()
plot_unpooled_signal = unpooled_data[0, 0, :].detach().numpy()

# 5. Create a plot to visualize all three signals
plt.figure(figsize=(16, 6))
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7, linestyle='-')
plt.plot(plot_pooled_signal, label='Max-Pooled 1D Signal', alpha=0.7, linestyle='--')
plt.plot(plot_unpooled_signal, label='Max-Unpooled 1D Signal', alpha=0.7, linestyle=':')
plt.title('Original, Max-Pooled, and Max-Unpooled 1D Signals')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)
plt.show()
```

To perform the average pooling operation, we will define a `torch.nn.AvgPool1d` layer, apply it to the previously generated `input_signal_1d`, and then plot both the input and the resulting pooled output to visualize the downsampling and smoothing effect.



```python
import torch.nn as nn
import matplotlib.pyplot as plt

# 1. Define a 1D Average Pooling layer
# Using kernel_size=2, stride=2 for downsampling
avgpool1d_layer = nn.AvgPool1d(kernel_size=2, stride=2)

# 2. Apply the average pooling layer to the generated input signal
output_avg_pooled_signal_1d = avgpool1d_layer(input_signal_1d)

# 3. Print the shapes of both input and output signals
print("Input signal shape:", input_signal_1d.shape)
print("Output signal shape after AvgPool1d:", output_avg_pooled_signal_1d.shape)

# 4. Extract the signals for plotting
plot_input_signal = input_signal_1d[0, 0, :].detach().numpy()
plot_output_avg_pooled_signal = output_avg_pooled_signal_1d[0, 0, :].detach().numpy()

# 5 & 6. Plot both the original and average-pooled signals
plt.figure(figsize=(14, 6))
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.plot(plot_output_avg_pooled_signal, label='Average-Pooled 1D Signal', alpha=0.7)
plt.title('Original vs. Average-Pooled 1D Signal (Downsampling and Smoothing Effect)')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)
plt.show()
```

## Effect of `kernel_size` on `nn.AvgPool1d`

This section illustrates how changing the `kernel_size` parameter in `torch.nn.AvgPool1d` influences the resulting signal. A larger kernel size means averaging over a wider window, leading to more aggressive downsampling and smoothing.

```python
import torch.nn as nn
import matplotlib.pyplot as plt

# Define AvgPool1d layers with different kernel sizes
# For simplicity, we'll set stride equal to kernel_size for non-overlapping pooling
avgpool1d_k2 = nn.AvgPool1d(kernel_size=2, stride=2)
avgpool1d_k10 = nn.AvgPool1d(kernel_size=10, stride=10)

# Apply average pooling with different kernel sizes to the generated input signal
output_avg_pooled_k2 = avgpool1d_k2(input_signal_1d)
output_avg_pooled_k10 = avgpool1d_k10(input_signal_1d)

print("Input signal shape:", input_signal_1d.shape)
print("Output shape with kernel_size=2:", output_avg_pooled_k2.shape)
print("Output shape with kernel_size=10:", output_avg_pooled_k10.shape)

# Extract signals for plotting
plot_input_signal = input_signal_1d[0, 0, :].detach().numpy()
plot_output_avg_pooled_k2 = output_avg_pooled_k2[0, 0, :].detach().numpy()
plot_output_avg_pooled_k10 = output_avg_pooled_k10[0, 0, :].detach().numpy()

# Plotting
plt.figure(figsize=(16, 12))

plt.subplot(3, 1, 1) # Three rows, one column, first plot
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.title('Original 1D Signal')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)

plt.subplot(3, 1, 2) # Three rows, one column, second plot
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.plot(plot_output_avg_pooled_k2, label='AvgPool1d (kernel_size=2) Output', alpha=0.7)
plt.title('Original vs. AvgPool1d (kernel_size=2) Output')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)

plt.subplot(3, 1, 3) # Three rows, one column, third plot
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.plot(plot_output_avg_pooled_k10, label='AvgPool1d (kernel_size=10) Output', alpha=0.7)
plt.title('Original vs. AvgPool1d (kernel_size=10) Output')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)

plt.tight_layout()
plt.show()
```

### Observation on `kernel_size` Effect in `nn.AvgPool1d`

From the plots, we can observe the following impacts of `kernel_size` on `nn.AvgPool1d`:

*   **Smaller `kernel_size` (e.g., 2)**: When `kernel_size=2` (with `stride=2`), the average pooling operation averages values over small windows. This results in a downsampled signal that retains more of the original signal's localized variations, though it is smoother than the original.

*   **Larger `kernel_size` (e.g., 10)**: When `kernel_size=10` (with `stride=10`), the average pooling operation averages values over much larger windows. This leads to more aggressive downsampling and significantly greater smoothing. The output signal becomes a much more abstract, compressed representation, effectively capturing the overall trend or low-frequency components of the original signal while discarding fine-grained details.

In summary, increasing the `kernel_size` in `nn.AvgPool1d` leads to more pronounced downsampling and smoothing, which can be useful for reducing noise, extracting global features, and further reducing the dimensionality of the signal.

## Apply and Plot LPPool1d

LPPool1D raises each value in the window to the power of p, finds their average, and then takes the p-th root of that sum. The parameter p can be chosen by you, such that:

    If p = 1, you get average pooling.
    If p approaches infinity, you get max pooling.

Why use LPPool1D?

You can set p to pick a pooling style that works best for your data, somewhere between averaging and taking the maximum.



To apply the LP pooling operation, we will define a `torch.nn.LPPool1d` layer, apply it to the `input_signal_1d`and print the shapes. Then we plot both the input and the resulting pooled output to visualize the downsampling and smoothing effect.



```python
import torch.nn as nn
import matplotlib.pyplot as plt

# 1. Define a 1D Power-Average Pooling layer
# Using norm_type=2, kernel_size=2, stride=2 for downsampling
lppool1d_layer = nn.LPPool1d(norm_type=2, kernel_size=2, stride=2)

# 2. Apply the LP pooling layer to the generated input signal
output_lp_pooled_signal_1d = lppool1d_layer(input_signal_1d)

# 3. Print the shapes of both input and output signals
print("Input signal shape:", input_signal_1d.shape)
print("Output signal shape after LPPool1d:", output_lp_pooled_signal_1d.shape)

# 4. Extract the signals for plotting
plot_input_signal = input_signal_1d[0, 0, :].detach().numpy()
plot_output_lp_pooled_signal = output_lp_pooled_signal_1d[0, 0, :].detach().numpy()

# 5 & 6. Plot both the original and LP-pooled signals
plt.figure(figsize=(14, 6))
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.plot(plot_output_lp_pooled_signal, label='LP-Pooled 1D Signal', alpha=0.7)
plt.title('Original vs. LP-Pooled 1D Signal (Downsampling and Smoothing Effect)')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)
plt.show()
```

## Apply and Plot AdaptiveMaxPool1d

Normally, when you use max pooling, you set the size of the pooling window. The output size depends on the input size and the window you choose.

With adaptive max pooling, you do the opposite: instead of picking a window size, you choose the exact output size you want, and PyTorch automatically figures out how to pool your data to fit that size.

Why use it?

* Great for models that need a fixed-size output, such as when preparing data for fully connected layers.
* Makes it easy to handle inputs of different sizes without needing to resize or crop them yourself.


To apply adaptive max pooling, we will define a `torch.nn.AdaptiveMaxPool1d` layer with a specific `output_size`, apply it to the `input_signal_1d`,and print the shapes to show the effect. Then we plot both the original and the adaptively pooled signals for visualization.



```python
import torch.nn as nn
import matplotlib.pyplot as plt

# 1. Define a 1D Adaptive Max Pooling layer
# Targeting an output size of 50
adaptive_maxpool1d_layer = nn.AdaptiveMaxPool1d(output_size=50)

# 2. Apply the adaptive max pooling layer to the generated input signal
output_adaptive_max_pooled_signal_1d = adaptive_maxpool1d_layer(input_signal_1d)

# 3. Print the shapes of both input and output signals
print("Input signal shape:", input_signal_1d.shape)
print("Output signal shape after AdaptiveMaxPool1d:", output_adaptive_max_pooled_signal_1d.shape)

# 4. Extract the signals for plotting
plot_input_signal = input_signal_1d[0, 0, :].detach().numpy()
plot_output_adaptive_max_pooled_signal = output_adaptive_max_pooled_signal_1d[0, 0, :].detach().numpy()

# 5. Create a plot to visualize both signals
plt.figure(figsize=(14, 6))
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.plot(plot_output_adaptive_max_pooled_signal, label='Adaptive Max-Pooled 1D Signal', alpha=0.7)
plt.title('Original vs. Adaptive Max-Pooled 1D Signal (Adaptive Downsampling Effect)')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)
plt.show()
```

## Apply and Plot ReflectionPad1d

Reflection padding increases the size of your data by adding extra values to the ends. Instead of filling these new values with zeros (zero-padding), it copies, or “reflects,” the values from the edge of your data. For example, if your data starts with [3, 5, 7] and you add reflection padding, the padding on the left might look like [5, 3, 3, 5, 7].

Why use reflection padding?

* Reduces edge effects: When neural networks process data, padding helps them handle the edges better. Reflection padding gives the network more natural context by mirroring real data, making processed results at the edges smoother.
* Keeps patterns intact: Instead of creating sharp, unnatural borders with zeros, reflection maintains the flow of patterns at the boundaries.

In the below block, using a padding size of 10, the output has 10 data added at the beginning and the end for a total size of 520.

```python
import torch.nn as nn
import matplotlib.pyplot as plt

# 1. Define a 1D Reflection Padding layer
# Using padding of 10
reflection_pad1d_layer = nn.ReflectionPad1d(padding=10)

# 2. Apply the reflection padding layer to the generated input signal
output_reflection_padded_signal_1d = reflection_pad1d_layer(input_signal_1d)

# 3. Print the shapes of both input and output signals
print("Input signal shape:", input_signal_1d.shape)
print("Output signal shape after ReflectionPad1d:", output_reflection_padded_signal_1d.shape)

# 4. Extract the signals for plotting
plot_input_signal = input_signal_1d[0, 0, :].detach().numpy()
plot_output_reflection_padded_signal = output_reflection_padded_signal_1d[0, 0, :].detach().numpy()

# 5. Create a plot to visualize both signals
plt.figure(figsize=(14, 6))
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.plot(plot_output_reflection_padded_signal, label='Reflection-Padded 1D Signal', alpha=0.7)
plt.title('Original vs. Reflection-Padded 1D Signal (Boundary Reflection Effect)')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)
plt.show()
```

## Apply and Plot ReplicationPad1d

Replication padding increases the size of your data by adding extra values to the ends. Instead of filling these new spots with zeros or mirrored values, it simply repeats the edge values of your data. For example, if your data is [3, 5, 7] and you add replication padding, you might get [3, 3, 3, 5, 7, 7, 7]—the start and end values are repeated to pad out the data.

Why use replication padding?

    Simple edge handling: Replicating the edge values gives your neural network extra context at the boundaries, and helps it process the ends of sequences without introducing new or strange values.
    Reduces unnatural effects: By not adding zeros or reflecting values that might not fit, replication keeps the input structure consistent at the borders.


```python
import torch.nn as nn
import matplotlib.pyplot as plt

# 1. Define a 1D Replication Padding layer
# Using padding of 5
replication_pad1d_layer = nn.ReplicationPad1d(padding=5)

# 2. Apply the replication padding layer to the generated input signal
output_replication_padded_signal_1d = replication_pad1d_layer(input_signal_1d)

# 3. Print the shapes of both input and output signals
print("Input signal shape:", input_signal_1d.shape)
print("Output signal shape after ReplicationPad1d:", output_replication_padded_signal_1d.shape)

# 4. Extract the signals for plotting
plot_input_signal = input_signal_1d[0, 0, :].detach().numpy()
plot_output_replication_padded_signal = output_replication_padded_signal_1d[0, 0, :].detach().numpy()

# 5. Create a plot to visualize both signals
plt.figure(figsize=(14, 6))
plt.plot(plot_input_signal, label='Original 1D Signal', alpha=0.7)
plt.plot(plot_output_replication_padded_signal, label='Replication-Padded 1D Signal', alpha=0.7)
plt.title('Original vs. Replication-Padded 1D Signal (Boundary Replication Effect)')
plt.xlabel('Length')
plt.ylabel('Value')
plt.legend()
plt.grid(False)
plt.show()
```

## Apply and Plot ConstantPad1d

ConstantPad1d increases the size of your data by adding extra values to the beginnings or ends of your sequence. Unlike other padding types, here you get to choose the exact value that fills those new spaces. For example, if you set the constant value to zero, the padded sequence might look like [0, 0, 3, 5, 7, 0, 0].
