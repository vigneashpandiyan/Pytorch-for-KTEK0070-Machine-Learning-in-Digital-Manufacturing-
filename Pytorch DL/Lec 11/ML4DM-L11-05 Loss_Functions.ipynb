# Loss Functions


Loss functions quantify how far a model’s predictions are from the true labels/targets. During training, the optimizer adjusts the model’s parameters to minimize this loss, which usually improves accuracy on the task. Different losses suit different problems, for example: for regression, MSE/MAE measure numeric error; for classification, cross-entropy measures how well predicted probabilities match the correct class. Choosing an appropriate loss helps training be stable and aligned with the metric you care about.

### L1 loss Function


1. nn.L1Loss - Creates a criterion that measures the mean absolute value of the element-wise difference between input x and target y.

Function:
    
$$\ell(x, y) = L = \{l_1,\dots,l_N\}^\top, \quad
l_n = \left| x_n - y_n \right|$$

```python
print("Module implementation")
loss = nn.L1Loss()
input_data = torch.randn(10, requires_grad=True)
target_data = torch.randn(10)
output_loss = loss(input_data, target_data)
output_loss.backward()    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.

print("Input Data = ", input_data)
print("Output Loss = ", output_loss)
print("\n")

print("Functional implementation")
output_loss = F.l1_loss(input_data, target_data, reduction='mean')
output_loss.backward()    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.
print("Output Loss = ", output_loss)
print("\n")

```

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import numpy as np

# 1. SETUP DATA
# Using a small batch so we can see every point clearly
input_data = torch.tensor([1.5, -0.5, 2.0, -1.0, 0.5], requires_grad=True)
target_data = torch.tensor([0.0,  0.0, 1.0, -2.0, 0.5])

# 2. CALCULATE LOSS (Forward Pass)
loss_fn = nn.L1Loss(reduction='none') # 'none' keeps the loss for each item separately
elementwise_loss = loss_fn(input_data, target_data)
final_loss = elementwise_loss.mean()

# 3. VISUALIZATION
input_np = input_data.detach().numpy()
target_np = target_data.numpy()
diff_np = elementwise_loss.detach().numpy()

plt.figure(figsize=(10, 6))
indices = np.arange(len(input_data))

# Plot Targets (Green X)
plt.scatter(indices, target_np, color='green', s=100, label='Ground Truth (Target)', marker='x', zorder=3)

# Plot Inputs (Blue Dots)
plt.scatter(indices, input_np, color='blue', s=100, label='Original Value (Input)', zorder=3)

# Plot the "Difference" (Red Lines)
for i in range(len(input_data)):
    # Draw line from input to target
    plt.plot([i, i], [input_np[i], target_np[i]], color='red', linestyle='--', linewidth=2, alpha=0.6)

    # Annotate the specific difference value
    mid_point = (input_np[i] + target_np[i]) / 2
    plt.text(i + 0.05, mid_point, f"Diff: {diff_np[i]:.1f}", color='red', fontsize=10)

# Add the final mean calculation visually
plt.axhline(final_loss.item(), color='purple', linestyle=':', alpha=0.5, label=f'Final L1 Loss (Mean): {final_loss.item():.2f}')

# Formatting
plt.title(f"Visualizing L1 Loss Calculation\nMean of Differences = {final_loss.item():.2f}")
plt.xlabel("Sample Index")
plt.ylabel("Value")
plt.xticks(indices, [f"Sample {i}" for i in indices])
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```

### Mean Square Error Loss

2. nn.MSELoss - Creates a criterion that measures the mean squared error between n elements in the input x and target y.
        
        
        
    Function:
$$\ell(x, y) = L = \{l_1,\dots,l_N\}^\top, \quad
l_n = \left( x_n - y_n \right)^2$$

```python
print("Module implementation")
loss = nn.MSELoss()
input_data = torch.randn(10, requires_grad=True)
target_data = torch.randn(10)
output_loss = loss(input_data, target_data)
output_loss.backward()    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.

print("Input Data = ", input_data)
print("Output Loss = ", output_loss)
print("\n")

print("Functional implementation")
output_loss = F.mse_loss(input_data, target_data, reduction='mean')
output_loss.backward()    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.
print("Output Loss = ", output_loss)
print("\n")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# 1. SETUP DATA (MSE Context)
# ==========================================
# Using small sample data to keep the plot readable
input_data = torch.tensor([1.5, -0.5, 2.0, -1.0, 0.5], requires_grad=True)
target_data = torch.tensor([0.0,  0.0, 1.0, -2.0, 0.5])

# ==========================================
# 2. CALCULATE MSE LOSS
# ==========================================
# 'none' gives us the squared error for each point: (input - target)^2
loss_fn = nn.MSELoss(reduction='none')
squared_errors = loss_fn(input_data, target_data)

# The final loss is the mean of these squares
final_loss = squared_errors.mean()

# ==========================================
# 3. VISUALIZATION
# ==========================================
input_np = input_data.detach().numpy()
target_np = target_data.numpy()
sq_err_np = squared_errors.detach().numpy()

plt.figure(figsize=(10, 6))
indices = np.arange(len(input_data))

# A. Plot Targets (Green X)
plt.scatter(indices, target_np, color='green', s=100, label='Ground Truth (Target)', marker='x', zorder=3)

# B. Plot Inputs (Blue Dots)
plt.scatter(indices, input_np, color='blue', s=100, label='Prediction (Input)', zorder=3)

# C. Plot the Error Bars and Annotate Squares
for i in range(len(input_data)):
    # 1. The visual line is just the linear distance (residual)
    plt.plot([i, i], [input_np[i], target_np[i]], color='red', linestyle='--', linewidth=2, alpha=0.6)

    # 2. The annotation shows the SQUARED value
    # We place the text at the midpoint of the line
    mid_point = (input_np[i] + target_np[i]) / 2

    # Calculate linear difference for display purposes
    linear_diff = input_np[i] - target_np[i]

    # Text: "Diff^2 = Value"
    label_text = f"({linear_diff:.1f})² = {sq_err_np[i]:.2f}"
    plt.text(i + 0.1, mid_point, label_text, color='darkred', fontsize=9, fontweight='bold')

# D. Show Final Loss
plt.axhline(final_loss.item(), color='purple', linestyle=':', alpha=0.5, label=f'Final MSE Loss: {final_loss.item():.2f}')

# Formatting
plt.title(f"Visualizing MSE Loss (L2)\nSum of Squares / N = {final_loss.item():.2f}")
plt.xlabel("Sample Index")
plt.ylabel("Value")
plt.xticks(indices, [f"Sample {i}" for i in indices])
plt.legend(loc='upper right')
plt.grid(True, alpha=0.3)
plt.tight_layout()

print("Input Data: ", input_data)
print("Squared Errors: ", squared_errors)
print("Final MSE Loss: ", final_loss)

plt.show()
```

### Cross Entropy Loss

3. nn.CrossEntropyLoss - This criterion combines nn.LogSoftmax() and nn.NLLLoss() in one single class. It is useful when training a classification problem with C classes. If provided, the optional argument weight should be a 1D Tensor assigning weight to each of the classes. This is particularly useful when you have an unbalanced training set.

Parameters:

    - weight (Tensor, optional) – a manual rescaling weight given to each class. If given, has to be a Tensor of size C

    - ignore_index (int, optional) – Specifies a target value that is ignored and does not contribute to the input gradient. When size_average is True, the loss is averaged over non-ignored targets.

    - reduction (string, optional) – Specifies the reduction to apply to the output: ‘none’ | ‘elementwise_mean’ | ‘sum’. ‘none’: no reduction will be applied, ‘elementwise_mean’: the sum of the output will be divided by the number of elements in the output, ‘sum’: the output will be summed. Default: ‘elementwise_mean’
    
Function:

$$ \text{loss}(x, class) = weight[class] \left(-x[class] + \log\left(\sum_j \exp(x[j])\right)\right) $$


```python
print("Module Implementation")
loss = nn.CrossEntropyLoss()
softmax = nn.Softmax()
input_data = torch.randn(3, 3, requires_grad=True)
print("Shape of input_data:", input_data.shape)
input_data = softmax(input_data)
print("Shape of input_data:", input_data.shape)
target_data = torch.tensor([0, 1, 2])
print("Shape of target_data:", target_data.shape)
output_loss = loss(input_data, target_data)
output_loss.backward(retain_graph=True)    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.

print("Input Data = ", input_data)
print("\n")
print("Target Data = ", target_data)
print("\n")
print("Output Loss = ", output_loss)
print("\n")
print("\n")


print("Functional implementation")
output_loss = F.cross_entropy(input_data, target_data)
output_loss.backward()    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.
print("Output Loss = ", output_loss)
print("\n")
```

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# ==========================================
# 1. SETUP DATA (Corrected Flow)
# ==========================================
# We define raw logits (no pre-softmax) for 3 samples, 3 classes
# Sample 1: Confident & Correct (Target 0)
# Sample 2: Confused (Target 1)
# Sample 3: Confident & Wrong (Target 2)
input_logits = torch.tensor([
    [5.0, 1.0, 0.5],   # Model strongly thinks Class 0
    [2.0, 2.0, 1.5],   # Model is unsure
    [3.0, 0.5, 0.5]    # Model strongly thinks Class 0 (But Target is 2!)
], requires_grad=True)

target_indices = torch.tensor([0, 1, 2])

# ==========================================
# 2. FORWARD & BACKWARD
# ==========================================
loss_fn = nn.CrossEntropyLoss(reduction='none') # 'none' to see loss per sample

# A. PyTorch applies Softmax internally here
output_loss = loss_fn(input_logits, target_indices)
output_loss.mean().backward()

# B. Manually calculate Probabilities for visualization
# (This is what CrossEntropy does inside)
probs = torch.softmax(input_logits, dim=1).detach().numpy()
grads = input_logits.grad.detach().numpy()
logits_np = input_logits.detach().numpy()
loss_np = output_loss.detach().numpy()

# ==========================================
# 3. VISUALIZATION
# ==========================================
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
classes = ['Class 0', 'Class 1', 'Class 2']
samples = ['Sample 1', 'Sample 2', 'Sample 3']

# --- Plot 1: Raw Inputs (Logits) ---
sns.heatmap(logits_np, annot=True, cmap="Blues", fmt=".1f", ax=axes[0,0],
            xticklabels=classes, yticklabels=samples)
axes[0,0].set_title("1. Raw Input (Logits)\n(Higher = Model Preference)")

# --- Plot 2: Probabilities (Softmax) ---
# Highlight the Target Class with a red box
sns.heatmap(probs, annot=True, cmap="Greens", fmt=".2f", ax=axes[0,1],
            xticklabels=classes, yticklabels=samples)
axes[0,1].set_title("2. Probabilities (Internal Softmax)\n(What the loss function sees)")

# Add boxes around targets to show what we are comparing against
for i, target in enumerate(target_indices):
    axes[0,1].add_patch(plt.Rectangle((target, i), 1, 1, fill=False, edgecolor='red', lw=4))

# --- Plot 3: The Resulting Loss ---
# Reshape for heatmap visualization (3x1)
loss_view = loss_np.reshape(-1, 1)
sns.heatmap(loss_view, annot=True, cmap="Reds", fmt=".3f", ax=axes[1,0],
            xticklabels=["Loss Value"], yticklabels=samples)
axes[1,0].set_title(f"3. Calculated Loss (-log(p))\nTotal Mean Loss: {loss_np.mean():.3f}")

# --- Plot 4: The Gradients (dLoss/dx) ---
# This shows how the model wants to change the inputs
sns.heatmap(grads, annot=True, cmap="coolwarm", center=0, fmt=".2f", ax=axes[1,1],
            xticklabels=classes, yticklabels=samples)
axes[1,1].set_title("4. Gradients (Backprop)\n(Blue=Increase Score, Red=Decrease Score)")

plt.tight_layout()
plt.show()
```

### Negative log likelihood loss

4. nn.NLLLoss - The negative log likelihood loss. It is useful to train a classification problem with 'C' number of classes. If provided, the optional argument weight should be a 1D Tensor assigning weight to each of the classes. This is particularly useful when you have an unbalanced training set.

Parameters:

    - weight (Tensor, optional) – a manual rescaling weight given to each class. If given, it has to be a Tensor of size C. Otherwise, it is treated as if having all ones.

    - ignore_index (int, optional) – Specifies a target value that is ignored and does not contribute to the input gradient. When size_average is True, the loss is averaged over non-ignored targets.

    - reduction (string, optional) – Specifies the reduction to apply to the output: ‘none’ | ‘elementwise_mean’ | ‘sum’. ‘none’: no reduction will be applied, ‘elementwise_mean’: the sum of the output will be divided by the number of elements in the output, ‘sum’: the output will be summed. Default: ‘elementwise_mean’

Function:

$$\begin{split}\ell(x, y) = \begin{cases}
    \sum_{n=1}^N \frac{1}{\sum_{n=1}^N w_{y_n}} l_n, & \text{if}\;
    \text{size_average} = \text{True},\\
    \sum_{n=1}^N l_n,  & \text{if}\;
    \text{size_average} = \text{False}.
\end{cases}\end{split}
$$

```python
print("Module Implementation")
m = nn.LogSoftmax()
loss = nn.NLLLoss()
# input is of size N x C = 3 x 5
input_data = torch.randn(3, 10, requires_grad=True)
input_data = m(input_data)
# each element in target has to have 0 <= value < C
target_data = torch.tensor([1, 0, 4])
output_loss = loss(input_data, target_data)
output_loss.backward(retain_graph=True)


print("Input Data = ", input_data)
print("\n")
print("Target Data = ", target_data)
print("\n")
print("Output Loss = ", output_loss)
print("\n")
print("\n")


print("Functional implementation")
output_loss = F.nll_loss(input_data, target_data)
output_loss.backward()    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.
print("Output Loss = ", output_loss)
print("\n")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# 1. SETUP & RUN YOUR LOGIC
# ==========================================
torch.manual_seed(42) # For reproducible plots

# Module Implementation
m = nn.LogSoftmax(dim=1)
loss_fn = nn.NLLLoss(reduction='none') # Use 'none' to see loss per sample

# Input: N=3 samples, C=10 classes
raw_logits = torch.randn(3, 10, requires_grad=True)
log_probs = m(raw_logits)  # These are values like -0.5, -2.3, etc.

target_data = torch.tensor([1, 0, 4]) # Targets for the 3 samples

# Calculate Loss
output_loss = loss_fn(log_probs, target_data)

# ==========================================
# 2. PREPARE DATA FOR PLOTTING
# ==========================================
# Convert Log-Probs back to standard Probabilities (0 to 1) for easier reading
probs = torch.exp(log_probs).detach().numpy()
targets = target_data.numpy()
loss_values = output_loss.detach().numpy()

# ==========================================
# 3. GENERATE PROBABILITY PLOTS
# ==========================================
fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharey=True)
classes = np.arange(10)

for i in range(3):
    ax = axes[i]
    sample_probs = probs[i]
    target_idx = targets[i]

    # 1. Plot all bars in neutral Blue
    bars = ax.bar(classes, sample_probs, color='lightgray', edgecolor='gray')

    # 2. Highlight the TARGET bar in Green (or Red if probability is low)
    # This specific bar determines the Loss!
    target_prob = sample_probs[target_idx]
    color = 'green' if target_prob > 0.5 else 'salmon'
    bars[target_idx].set_color(color)
    bars[target_idx].set_edgecolor('black')
    bars[target_idx].set_linewidth(1.5)

    # 3. Annotations
    ax.set_title(f"Sample {i+1} (Target: {target_idx})")
    ax.set_xlabel("Class Index")
    if i == 0: ax.set_ylabel("Probability")

    # Add text above the target bar showing the NLL Loss contribution
    # NLL Loss is literally: -1 * log_probs[target]
    height = sample_probs[target_idx]
    ax.text(target_idx, height + 0.02, f"Loss\n{loss_values[i]:.2f}",
            ha='center', va='bottom', fontweight='bold', color='darkred')

    # Add text for the probability
    ax.text(target_idx, height/2, f"{height:.2f}",
            ha='center', va='center', color='white', fontweight='bold')

plt.suptitle("Visualizing NLLLoss: The lower the Target Probability, the higher the Loss", fontsize=14)
plt.tight_layout()
plt.show()
```

### Poisson Negative log likelihood loss

5. nn.PoissonNLLLoss - Negative log likelihood loss with Poisson distribution of target.

Parameters:

        - log_input (bool, optional) – if True, the loss is computed as exp(input)−target∗input, if False the loss is input−target∗log(input+eps)

        - full (bool, optional) – whether to compute full loss, i. e. to add the Stirling approximation term target∗log(target)−target+0.5∗log(2πtarget).

        - eps (float, optional) – Small value to avoid evaluation of log(0), when log_input == False. Default: 1e-8

        - reduction (string, optional) – Specifies the reduction to apply to the output: ‘none’ | ‘elementwise_mean’ | ‘sum’. ‘none’: no reduction will be applied, ‘elementwise_mean’: the sum of the output will be divided by the number of elements in the output, ‘sum’: the output will be summed. Default: ‘elementwise_mean’
        
Function:

$$\begin{align}\begin{aligned}\text{target} \sim \mathrm{Poisson}(\text{input})\\\text{loss}(\text{input}, \text{target}) = \text{input} - \text{target} * \log(\text{input})
                            + \log(\text{target!})\end{aligned}\end{align}$$


```python
print("Module Implementation")
loss = nn.PoissonNLLLoss()
input_data = torch.randn(5, 2, requires_grad=True)
target_data = torch.randn(5, 2)
output_loss = loss(input_data, target_data)
output_loss.backward(retain_graph=True)


print("Input Data = ", input_data)
print("\n")
print("Target Data = ", target_data)
print("\n")
print("Output Loss = ", output_loss)
print("\n")
print("\n")

print("Functional implementation")
output_loss = F.poisson_nll_loss(input_data, target_data)
output_loss.backward()    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.
print("Output Loss = ", output_loss)
print("\n")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# 1. SETUP & RUN YOUR LOGIC
# ==========================================
torch.manual_seed(42) # For reproducible plots

# Module Definitions
loss_fn = nn.KLDivLoss(reduction='none') # 'none' to see loss per element
log_softmax = nn.LogSoftmax(dim=1)
softmax = nn.Softmax(dim=1)

batch_size = 2
classes = 3

# Create Data
input_data = torch.randn(batch_size, classes, requires_grad=True)
target_data = torch.randn(batch_size, classes, requires_grad=True)

# Apply Activations
# Q (Prediction) must be Log-Probabilities for PyTorch KLDivLoss
log_input_Q = log_softmax(input_data)
# P (Target) must be Probabilities
prob_target_P = softmax(target_data)

# Calculate Loss
# Formula: P * (log(P) - log(Q))  [Essentially P * log(P/Q)]
elementwise_loss = loss_fn(log_input_Q, prob_target_P)
# Sum over classes to get KL per sample
sample_losses = elementwise_loss.sum(dim=1).detach().numpy()

# ==========================================
# 2. PREPARE DATA FOR PLOTTING
# ==========================================
# We need both P and Q in "Probability" format (0.0 to 1.0) to compare them visually
P_probs = prob_target_P.detach().numpy()            # Target Distribution
Q_probs = torch.exp(log_input_Q).detach().numpy()   # Predicted Distribution
class_indices = np.arange(classes)

# ==========================================
# 3. GENERATE COMPARISON PLOTS
# ==========================================
fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharey=True)

for i in range(batch_size):
    ax = axes[i]

    # Plot Target (P) - The "Gold Standard" we want to match
    ax.bar(class_indices - 0.2, P_probs[i], width=0.4, label='Target (P)', color='green', alpha=0.7)

    # Plot Prediction (Q) - The Approximation
    ax.bar(class_indices + 0.2, Q_probs[i], width=0.4, label='Prediction (Q)', color='blue', alpha=0.7)

    # Formatting
    ax.set_title(f"Sample {i+1}\nKL Divergence: {sample_losses[i]:.4f}")
    ax.set_xticks(class_indices)
    ax.set_xlabel("Class Index")
    ax.set_ylim(0, 1.0)
    ax.grid(True, alpha=0.2)

    if i == 0:
        ax.set_ylabel("Probability")
        ax.legend()

    # Add text annotation
    # KL is low when Blue bars match Green bars height
    status = "Good Match" if sample_losses[i] < 0.1 else "Mismatch"
    ax.text(1, 0.9, status, ha='center', fontsize=12, fontweight='bold',
            color='green' if status == "Good Match" else 'red')

plt.suptitle(f"Visualizing KL Divergence: Matching Distributions (P vs Q)", fontsize=14)
plt.tight_layout()
plt.show()
```

### Smooth L1 Loss

12. nn.SmoothL1Loss - Creates a criterion that uses a squared term if the absolute element-wise error falls below 1 and an L1 term otherwise. It is less sensitive to outliers than the MSELoss and in some cases prevents exploding gradients (e.g. see “Fast R-CNN” paper by Ross Girshick).

Parameter:

        - reduction (string, optional) – Specifies the reduction to apply to the output: ‘none’ | ‘elementwise_mean’ | ‘sum’. ‘none’: no reduction will be applied, ‘elementwise_mean’: the sum of the output will be divided by the number of elements in the output, ‘sum’: the output will be summed. Default: ‘elementwise_mean’

Function:

$$\text{loss}(x, y) = \frac{1}{n} \sum_{i} z_{i}$$

Where, $$z_{i}\text{ is }$$

\begin{split}z_{i} =
\begin{cases}
0.5 (x_i - y_i)^2, & \text{if } |x_i - y_i| < 1 \\
|x_i - y_i| - 0.5, & \text{otherwise }
\end{cases}\end{split}

```python
print("Module implementation")
loss = nn.SmoothL1Loss()
input_data = torch.randn(3, requires_grad=True)
target_data = torch.randn(3, requires_grad=False)
output_loss = loss(input_data, target_data)
output_loss.backward(retain_graph=True)    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.

print("Input Data = ", input_data)
print("\n")
print("Target Data = ", target_data)
print("\n")
print("Output Loss = ", output_loss)
print("\n")
print("\n")

print("Functional implementation")
output_loss = F.smooth_l1_loss(input_data, target_data)
output_loss.backward()    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.
print("Output Loss = ", output_loss)
print("\n")
```

### Multi Label Soft Margin loss

13. nn.MultiLabelSoftMarginLoss - Creates a criterion that optimizes a multi-label one-versus-all loss based on max-entropy, between input x and target y of size (N, C)

        - weight (Tensor, optional) – a manual rescaling weight given to each class. If given, it has to be a Tensor of size C. Otherwise, it is treated as if having all ones.
        
        - reduction (string, optional) – Specifies the reduction to apply to the output: ‘none’ | ‘elementwise_mean’ | ‘sum’. ‘none’: no reduction will be applied, ‘elementwise_mean’: the sum of the output will be divided by the number of elements in the output, ‘sum’: the output will be summed. Default: ‘elementwise_mean’



Function:

$$loss(x, y) = - \sum_i y[i] * \log((1 + \exp(-x[i]))^{-1})
                 + (1-y[i]) * \log\left(\frac{\exp(-x[i])}{(1 + \exp(-x[i]))}\right)
$$

```python
print("Module implementation")
loss = nn.MultiLabelSoftMarginLoss()
input_data = torch.randn((1, 3), requires_grad=True)
target_data = torch.randn((1, 3), requires_grad=True)
output_loss = loss(input_data, target_data)
output_loss.backward(retain_graph = True)    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.

print("Input Data = ", input_data)
print("\n")
print("Target Data = ", target_data)
print("\n")
print("Output Loss = ", output_loss)
print("\n")
print("\n")

print("Functional implementation")
output_loss = F.multilabel_soft_margin_loss(input_data, target_data)
output_loss.backward()    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.
print("Output Loss = ", output_loss)
print("\n")
```

### Cosine Embedding Loss

14. nn.CosineEmbeddingLoss - Creates a criterion that measures the loss given input tensors x1, x2 and a Tensor label y with values 1 or -1. This is used for measuring whether two inputs are similar or dissimilar, using the cosine distance, and is typically used for learning nonlinear embeddings or semi-supervised learning.

Parameter:

        - margin (float, optional) – Should be a number from -1 to 1, 0 to 0.5 is suggested. If margin is missing, the default value is 0.
       
        - reduction (string, optional) – Specifies the reduction to apply to the output: ‘none’ | ‘elementwise_mean’ | ‘sum’. ‘none’: no reduction will be applied, ‘elementwise_mean’: the sum of the output will be divided by the number of elements in the output, ‘sum’: the output will be summed. Default: ‘elementwise_mean’


Function:

$$\begin{split}\text{loss}(x, y) =
\begin{cases}
1 - \cos(x_1, x_2), & \text{if } y == 1 \\
\max(0, \cos(x_1, x_2) - \text{margin}), & \text{if } y == -1
\end{cases}\end{split}
$$

```python
print("Module Implementation")
loss = nn.CosineEmbeddingLoss()

input_data1 = torch.randn(3, 3, requires_grad=True)
input_data2 = torch.randn(3, 3, requires_grad=True)
target_data = torch.tensor([-1., 1., 1.])
output_loss = loss(input_data1, input_data2, target_data)
output_loss.backward(retain_graph = True)    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.

print("Input Data1 = ", input_data1)
print("\n")
print("Input Data2 = ", input_data2)
print("\n")
print("Target Data = ", target_data)
print("\n")
print("Output Loss = ", output_loss)
print("\n")
print("\n")

print("Functional implementation")
output_loss = F.cosine_embedding_loss(input_data1, input_data2, target_data)
output_loss.backward()    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.
print("Output Loss = ", output_loss)
print("\n")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# 1. SETUP DATA
# ==========================================
torch.manual_seed(42) # Fix seed for reproducible random vectors

# 3 Samples, 3 Dimensions each
input1 = torch.randn(3, 3, requires_grad=True)
input2 = torch.randn(3, 3, requires_grad=True)

# Targets:
# Sample 0: Target -1 (Should be different)
# Sample 1: Target  1 (Should be similar)
# Sample 2: Target  1 (Should be similar)
targets = torch.tensor([-1., 1., 1.])

# ==========================================
# 2. CALCULATE METRICS
# ==========================================
loss_fn = nn.CosineEmbeddingLoss(margin=0.0, reduction='none')

# A. Calculate actual Loss
output_loss = loss_fn(input1, input2, targets)
output_loss.sum().backward()

# B. Calculate Cosine Similarity for the X-Axis of the plot
# This tells us the current angle between the vectors (-1 to 1)
cos_sims = F.cosine_similarity(input1, input2)

# Convert to numpy for plotting
cos_np = cos_sims.detach().numpy()
loss_np = output_loss.detach().numpy()
targets_np = targets.numpy()

# ==========================================
# 3. VISUALIZATION
# ==========================================
plt.figure(figsize=(10, 6))

# --- A. Plot Theoretical Curves ---
# Generate a range of cosine similarities from -1 to 1
x_cos = np.linspace(-1, 1, 100)

# Curve for Target = 1: Loss = 1 - cos(theta)
y_pos = 1 - x_cos
plt.plot(x_cos, y_pos, color='green', linestyle='--', alpha=0.4, label='Target=1 (Align!)')

# Curve for Target = -1: Loss = max(0, cos(theta) - margin)
# Margin is 0.0 by default
y_neg = np.maximum(0, x_cos - 0.0)
plt.plot(x_cos, y_neg, color='red', linestyle='--', alpha=0.4, label='Target=-1 (Separate!)')


# --- B. Plot Your Data Points ---
for i in range(3):
    c_sim = cos_np[i]
    loss_val = loss_np[i]
    t = targets_np[i]

    if t == 1.0:
        # Target 1 (Green)
        plt.scatter(c_sim, loss_val, color='green', s=150, edgecolors='black', zorder=5)
        plt.text(c_sim, loss_val + 0.1, f"Sample {i}\n(T=1)", ha='center', color='green', fontweight='bold')
    else:
        # Target -1 (Red)
        plt.scatter(c_sim, loss_val, color='red', s=150, marker='s', edgecolors='black', zorder=5)
        plt.text(c_sim, loss_val + 0.1, f"Sample {i}\n(T=-1)", ha='center', color='red', fontweight='bold')

# Formatting
plt.title(f"Visualizing CosineEmbeddingLoss (Margin=0.0)")
plt.xlabel("Cosine Similarity (1.0 = Aligned, 0.0 = Orthogonal, -1.0 = Opposite)")
plt.ylabel("Loss Value")
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='gray', linestyle=':', label='Orthogonal Line')
plt.xlim(-1.1, 1.1)
plt.legend()
plt.grid(True, alpha=0.3)

# Helper Text
plt.text(0.8, 1.5, "Vectors pointing\nsame way", ha='center', fontsize=8, alpha=0.6)
plt.text(-0.8, 1.5, "Vectors pointing\nopposite way", ha='center', fontsize=8, alpha=0.6)

plt.show()
```

### Multi margin loss

15. nn.MultiMarginLoss - Creates a criterion that optimizes a multi-class classification hinge loss (margin-based loss) between input x (a 2D mini-batch Tensor) and output y (which is a 1D tensor of target class indices, 0≤y≤x.size(1))

Parameter:

        - p (int, optional) – Has a default value of 1. 1 and 2 are the only supported values

        - margin (float, optional) – Has a default value of 1.

        - weight (Tensor, optional) – a manual rescaling weight given to each class. If given, it has to be a Tensor of size C. Otherwise, it is treated as if having all ones.
        
        - reduction (string, optional) – Specifies the reduction to apply to the output: ‘none’ | ‘elementwise_mean’ | ‘sum’. ‘none’: no reduction will be applied, ‘elementwise_mean’: the sum of the output will be divided by the number of elements in the output, ‘sum’: the output will be summed. Default: ‘elementwise_mean’


Function:
    
$$\text{loss}(x, y) = \frac{\sum_i \max(0, \text{margin} - x[y] + x[i]))^p}{\text{x.size}(0)}$$

```python
print("Model Implementation")
loss = nn.MultiMarginLoss()
input_data = torch.randn(3, requires_grad=True) # Scores for 3 classes for a single sample
target_data = torch.tensor([1], requires_grad=False) # Target class index (e.g., class 1) for the single sample
output_loss = loss(input_data, target_data)
output_loss.backward(retain_graph = True)    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.

print("Input Data = ", input_data)
print("\n")
print("Target Data = ", target_data)
print("\n")
print("Output Loss = ", output_loss)
print("\n")
print("\n")

print("Functional implementation")
output_loss = F.multi_margin_loss(input_data, target_data)
output_loss.backward()    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.
print("Output Loss = ", output_loss)
print("\n")
```

### Triplet Margin Loss

16. nn.TripletMarginLoss - Creates a criterion that measures the triplet loss given an input tensors x1, x2, x3 and a margin with a value greater than 0. This is used for measuring a relative similarity between samples.

Parameter:

        - margin (float, optional) – Default: 1.

        - p (int, optional) – The norm degree for pairwise distance. Default: 2.

        - swap (float, optional) – The distance swap is described in detail in the paper 'Learning shallow convolutional feature descriptors with triplet losses' by V. Balntas, E. Riba et al. Default: False.

        - reduction (string, optional) – Specifies the reduction to apply to the output: ‘none’ | ‘elementwise_mean’ | ‘sum’. ‘none’: no reduction will be applied, ‘elementwise_mean’: the sum of the output will be divided by the number of elements in the output, ‘sum’: the output will be summed. Default: ‘elementwise_mean’


Function:
    
$$L(a, p, n) = \max \{d(a_i, p_i) - d(a_i, n_i) + {\rm margin}, 0\}
\text{where,}
d(x_i, y_i) = \left\lVert {\bf x}_i - {\bf y}_i \right\rVert_p$$

```python
print("Module implementation")
triplet_loss = nn.TripletMarginLoss(margin=1.0, p=2)
input_data1 = torch.randn(100, 128, requires_grad=True)
input_data2 = torch.randn(100, 128, requires_grad=True)
input_data3 = torch.randn(100, 128, requires_grad=True)
output = triplet_loss(input_data1, input_data2, input_data3)
output.backward(retain_graph = True)

print("Output Loss = ", output_loss)
print("\n")

print("Functional implementation")
output_loss = F.triplet_margin_loss(input_data1, input_data2, input_data3, margin=1.0, p=2)
output_loss.backward()    # loss.backward() computes dloss/dx for every parameter x which has requires_grad=True.
print("Output Loss = ", output_loss)
print("\n")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# 1. SETUP DATA
# ==========================================
# We use 100 samples of 128-dim vectors
anchor = torch.randn(100, 128)
positive = torch.randn(100, 128) # In real life, this would be closer to anchor
negative = torch.randn(100, 128)

margin = 1.0

# ==========================================
# 2. CALCULATE DISTANCES & LOSS
# ==========================================
# A. Calculate Euclidean distances manually for plotting
# PairwiseDistance(p=2) calculates the Euclidean norm
pdist = nn.PairwiseDistance(p=2)
dist_pos = pdist(anchor, positive).numpy() # x-axis
dist_neg = pdist(anchor, negative).numpy() # y-axis

# B. Calculate Actual Loss
# Loss = max(0, dist_pos - dist_neg + margin)
losses = F.triplet_margin_loss(anchor, positive, negative, margin=margin, reduction='none')
losses_np = losses.detach().numpy()

# ==========================================
# 3. VISUALIZATION (Distance Map)
# ==========================================
plt.figure(figsize=(10, 8))

# --- Define Zones ---
# We categorize triplets based on their difficulty
# 1. HARD: Negative is closer than Positive (d_neg < d_pos)
# 2. SEMI-HARD: Negative is further, but inside margin (d_pos < d_neg < d_pos + margin)
# 3. EASY: Negative is far enough (d_neg > d_pos + margin) -> Loss is 0

max_val = max(np.max(dist_pos), np.max(dist_neg)) + 2

# Draw the boundary lines
x = np.linspace(0, max_val, 100)
plt.plot(x, x, 'k--', label='Equal Distance (d_pos = d_neg)')
plt.plot(x, x + margin, 'r-', linewidth=2, label=f'Margin Boundary (d_neg = d_pos + {margin})')

# Shade the regions
# Region 1: Hard (Below diagonal)
plt.fill_between(x, 0, x, color='red', alpha=0.1)
plt.text(max_val*0.8, max_val*0.4, "HARD\n(Negative is closer!)", color='darkred', ha='center', fontsize=12, fontweight='bold')

# Region 2: Semi-Hard (Between lines)
plt.fill_between(x, x, x + margin, color='orange', alpha=0.1)
plt.text(max_val*0.1, max_val*0.3 + margin, "SEMI-HARD\n(Inside Margin)", color='darkorange', ha='center', fontsize=10, fontweight='bold', rotation=45)

# Region 3: Easy (Above margin line)
plt.fill_between(x, x + margin, max_val + 5, color='green', alpha=0.1)
plt.text(max_val*0.2, max_val*0.8, "EASY\n(Loss = 0)", color='green', ha='center', fontsize=12, fontweight='bold')


# --- Plot Points ---
# Color points by their loss value
sc = plt.scatter(dist_pos, dist_neg, c=losses_np, cmap='coolwarm', s=50, edgecolors='k', zorder=5)
cbar = plt.colorbar(sc)
cbar.set_label('Loss Value')

# Formatting
plt.title(f"Visualizing Triplet Loss Zones\nMargin = {margin}")
plt.xlabel("Positive Distance (Anchor ↔ Positive)\n(We want this SMALL ←)")
plt.ylabel("Negative Distance (Anchor ↔ Negative)\n(We want this LARGE ↑)")
plt.xlim(0, max_val)
plt.ylim(0, max_val)
plt.legend(loc='upper right')
plt.grid(True, alpha=0.3)

plt.show()
```
