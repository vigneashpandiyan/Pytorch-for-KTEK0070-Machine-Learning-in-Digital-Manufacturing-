# Inception

An Inception block is a CNN module that processes the same input through multiple parallel branches (different receptive fields), then concatenates the results along the channel dimension. This helps the network capture multi‑scale features (small and large patterns) at the same time.

## Load MNIST Dataset




```python
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# 2. Define a transformation pipeline
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,))
])

# 3. Download and load the MNIST training dataset
train_dataset = datasets.MNIST('./data', train=True, download=True, transform=transform)
print("MNIST training dataset loaded successfully.")

# 4. Download and load the MNIST test dataset
test_dataset = datasets.MNIST('./data', train=False, download=True, transform=transform)
print("MNIST test dataset loaded successfully.")

# 5. Create a DataLoader for the training dataset
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
print(f"Train DataLoader created with batch size 64 and {len(train_loader)} batches.")

# 6. Create a DataLoader for the test dataset
test_loader = DataLoader(test_dataset, batch_size=1000, shuffle=False)
print(f"Test DataLoader created with batch size 1000 and {len(test_loader)} batches.")

# 7. Determine the device to use for training
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

## SimpleInceptionNet Class


We will define a `SimpleInceptionBlock` with two parallel convolutional layers (e.g., a 1x1 convolution and a 3x3 convolution), and then create a `SimpleInceptionNet` that stacks these blocks. This will provide a two layered simple inception architecture.

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

# Define a SimpleInceptionBlock
class SimpleInceptionBlock(nn.Module):
    def __init__(self, in_channels, out_channels):
        super(SimpleInceptionBlock, self).__init__()
        self.branch1x1 = nn.Conv2d(in_channels, out_channels // 2, kernel_size=1)
        self.branch3x3 = nn.Conv2d(in_channels, out_channels // 2, kernel_size=3, padding=1)

    def forward(self, x):
        branch1x1 = self.branch1x1(x)
        branch3x3 = self.branch3x3(x)
        return torch.cat([branch1x1, branch3x3], 1)

# Define a SimpleInceptionNet class
class SimpleInceptionNet(nn.Module):
    def __init__(self, num_blocks, base_channels, num_classes=10):
        super(SimpleInceptionNet, self).__init__()

        # Enforce the 16-filter limit if desired (can be adjusted for Inception)
        if base_channels > 16:
            base_channels = 16
            print(f"Warning: base_channels adjusted to {base_channels} to respect the 16-filter limit.")

        self.conv1 = nn.Conv2d(1, base_channels, kernel_size=3, stride=1, padding=1)
        self.relu = nn.ReLU(inplace=True)
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)

        self.layers = nn.ModuleList()
        current_channels = base_channels
        for _ in range(num_blocks):
            # Ensure output channels for inception block are even for splitting
            out_c = current_channels if current_channels % 2 == 0 else current_channels + 1
            self.layers.append(SimpleInceptionBlock(current_channels, out_c))
            self.layers.append(nn.ReLU(inplace=True))
            current_channels = out_c

        self.avg_pool = nn.AdaptiveAvgPool2d((1, 1)) # Corrected from nn.AdaptiveAvg2d
        self.fc = nn.Linear(current_channels, num_classes)

    def forward(self, x):
        x = self.pool(self.relu(self.conv1(x)))
        for layer in self.layers:
            x = layer(x)
        x = self.avg_pool(x)
        x = torch.flatten(x, 1)
        x = self.fc(x)
        return x

print("SimpleInceptionBlock and SimpleInceptionNet classes defined successfully.")
```

## Instantiate Simple Inception Model (3 blocks, 16 filters)


We will create an instance of `SimpleInceptionNet` with the specified parameters and transfer it to the 'device' variable, similar to how the `SimpleResNet` was handled in the previous module.

```python
simple_inception_model = SimpleInceptionNet(num_blocks=3, base_channels=16)
simple_inception_model.to(device)
print(f"SimpleInceptionNet model instantiated with 3 blocks and 16 base channels, and moved to {device}.")
```

## Define Loss Function and Optimizer for Inception Model


We will instantiate `CrossEntropyLoss` as the criterion and `SGD` as the optimizer, passing the new model's parameters.

```python
import torch.optim as optim

# Define the loss function
criterion = nn.CrossEntropyLoss()

# Define the optimizer with SGD, model parameters, learning rate, and momentum
optimizer = optim.SGD(simple_inception_model.parameters(), lr=0.01, momentum=0.9)

print("Loss function (CrossEntropyLoss) and optimizer (SGD) defined successfully for Inception model.")
```

## Show Simple Inception Model Parameters



```python
from torchsummary import summary

# Display the model summary
summary(simple_inception_model, input_size=(1, 28, 28))
```

## Training

```python
train_losses = []
train_accuracies = []
num_epochs = 10 # Changed to 10 epochs for better training

print("Starting training...")

for epoch in range(num_epochs):
    simple_inception_model.train() # Set the model to training mode
    running_loss = 0.0
    correct_predictions = 0
    total_samples = 0

    for i, (images, labels) in enumerate(train_loader):
        images, labels = images.to(device), labels.to(device)

        # Zero the parameter gradients
        optimizer.zero_grad()

        # Forward pass
        outputs = simple_inception_model(images)
        loss = criterion(outputs, labels)

        # Backward pass and optimize
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

        # Calculate training accuracy
        _, predicted = torch.max(outputs.data, 1)
        total_samples += labels.size(0)
        correct_predictions += (predicted == labels).sum().item()

    epoch_loss = running_loss / len(train_loader)
    epoch_accuracy = 100 * correct_predictions / total_samples

    train_losses.append(epoch_loss)
    train_accuracies.append(epoch_accuracy)

    print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {epoch_loss:.4f}, Accuracy: {epoch_accuracy:.2f}%')

print("Training finished.")
```

## Saving and loading the model



```python
model_save_path = 'simple_inception_mnist.pth'
torch.save(simple_inception_model.state_dict(), model_save_path)
print(f"Simple Inception model state dictionary saved to {model_save_path}")
```

### Note
*   The saved model can now be easily reloaded in future sessions for inference or further training without needing to retrain it from scratch.
*   Colab will NOT store the model once runtime disconnects. It is highly recommended to download it to your local disk, if you want to upload it again.
*   Consider adding versioning or timestamps to the model filenames if multiple iterations or experiments are expected, to better manage different saved models.



```python
loaded_simple_model = SimpleInceptionNet(num_blocks=3, base_channels=16)
loaded_simple_model.load_state_dict(torch.load(model_save_path))
loaded_simple_model.to(device)
print(f"Simple Inception model loaded from {model_save_path} and moved to {device}.")
```

## Testing




To evaluate the model's performance on the test dataset, we implement an evaluation loop that iterates through the test data, calculates predictions, and accumulates loss and accuracy without computing gradients.



```python
print("Starting evaluation of the loaded model...")

loaded_simple_model.eval() # Set the model to evaluation mode
test_loss = 0
correct_predictions = 0
total_samples = 0

with torch.no_grad(): # Disable gradient calculation
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)

        outputs = loaded_simple_model(images)
        loss = criterion(outputs, labels)
        test_loss += loss.item()

        _, predicted = torch.max(outputs.data, 1)
        total_samples += labels.size(0)
        correct_predictions += (predicted == labels).sum().item()

test_loss /= len(test_loader) # Calculate average test loss
test_accuracy = 100 * correct_predictions / total_samples # Calculate test accuracy

print(f'Test Loss: {test_loss:.4f}, Test Accuracy: {test_accuracy:.2f}%')
print("Evaluation finished.")
```

## Plot Training Curves



```python
import matplotlib.pyplot as plt

# Create a figure with two subplots
plt.figure(figsize=(12, 5))

# Subplot 1: Training Loss
plt.subplot(1, 2, 1) # 1 row, 2 columns, 1st plot
plt.plot(range(1, num_epochs + 1), train_losses, label='Training Loss', color='blue')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training Loss Over Epochs')
plt.legend()
plt.grid(False)

# Subplot 2: Training Accuracy
plt.subplot(1, 2, 2) # 1 row, 2 columns, 2nd plot
plt.plot(range(1, num_epochs + 1), train_accuracies, label='Training Accuracy', color='green')
plt.xlabel('Epoch')
plt.ylabel('Accuracy (%)')
plt.title('Training Accuracy Over Epochs')
plt.legend()
plt.grid(False)

# Adjust layout to prevent overlapping titles and labels
plt.tight_layout()

# Display the plots
plt.show()

print("Training loss and accuracy curves plotted successfully.")
```

## Confusion Matrix



To generate the confusion matrix, we need to first collect all true labels and predicted labels from the test dataset. Then we will use `sklearn.metrics` to compute the confusion matrix and `seaborn` along with `matplotlib` to visualize it.



```python
import numpy as np
from sklearn.metrics import confusion_matrix
import seaborn as sns

# Initialize lists to store true and predicted labels
all_labels = []
all_predictions = []

# Set the model to evaluation mode
loaded_simple_model.eval()

# Disable gradient calculations
with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)

        outputs = loaded_simple_model(images)
        _, predicted = torch.max(outputs.data, 1)

        all_labels.extend(labels.cpu().numpy())
        all_predictions.extend(predicted.cpu().numpy())

# Convert lists to numpy arrays
all_labels = np.array(all_labels)
all_predictions = np.array(all_predictions)

# Compute the confusion matrix
cm = confusion_matrix(all_labels, all_predictions)

# Plot the confusion matrix
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=range(10), yticklabels=range(10))
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix for Simple ResNet on MNIST Test Set')
plt.show()

print("Confusion matrix generated and displayed successfully.")
```
